```python
import numpy as np
import joblib
from sklearn.ensemble import IsolationForest
```

```python
N_SAMPLES = 10000
RANDOM_STATE = 42

def generate_normal_traffic(n: int):
  """
  generates n samples of normal network traffic.
  features: [bytes, duration, src_port, dst_port]
  these ranges must match what make_normal_event() produces in fake_network_event.py
  """

  rng = np.random.default_rng(RANDOM_STATE)

  bytes_col = rng.integers(64, 1500, size=n).astype(float)
  duration_col = rng.uniform(0.01, 2.0, size=n)
  src_port_col = rng.integers(1024, 65535, size=n).astype(float)
  dst_port_col = rng.choice([80, 443, 22, 3306, 5432, 6379, 8080, 8443], size=n).astype(float)

  return np.column_stack([bytes_col, duration_col, src_port_col, dst_port_col])
```

```python
X = generate_normal_traffic(N_SAMPLES)
model = IsolationForest(
    n_estimators=100,
    contamination= 0.01,
    random_state=RANDOM_STATE
)
model = model.fit(X)
```
